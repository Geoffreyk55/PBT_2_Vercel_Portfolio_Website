````python
# Cell 13 — Static Artifact Validation

required_files = ["index.html", "styles.css", "script.js", "README.md", ".gitignore", "vercel.json"]
artifact_checks = {name: (site_dir / name).exists() and (site_dir / name).stat().st_size > 0 for name in required_files}

for name, status in artifact_checks.items():
    print(f"{name:<16}: {'PASSED' if status else 'FAILED'}")

static_artifacts_ready = all(artifact_checks.values())
print("Static website artifacts:", "READY" if static_artifacts_ready else "NOT READY")
````

````python
# Cell 14 — HTML Structure Validation

html_text = (site_dir / "index.html").read_text(encoding="utf-8")

required_markers = [
    "<!DOCTYPE html>",
    'name="viewport"',
    'id="about"',
    'id="skills"',
    'id="projects"',
    'id="program"',
    'id="contact"',
    "styles.css",
    "script.js",
]

html_checks = {marker: marker in html_text for marker in required_markers}

for marker, status in html_checks.items():
    print(f"{marker:<24}: {'PASSED' if status else 'FAILED'}")

html_structure_ready = all(html_checks.values())
print("HTML structure:", "READY" if html_structure_ready else "NOT READY")
````

````python
# Cell 15 — Responsive & Accessibility Validation

css_text = (site_dir / "styles.css").read_text(encoding="utf-8")

design_checks = {
    "responsive_grid": "auto-fit" in css_text,
    "mobile_media_query": "@media" in css_text,
    "responsive_typography": "clamp(" in css_text,
    "html_language": '<html lang="en">' in html_text,
    "semantic_main": "<main>" in html_text,
    "semantic_nav": "<nav>" in html_text,
    "semantic_footer": "<footer>" in html_text,
}

for check, status in design_checks.items():
    print(f"{check:<24}: {'PASSED' if status else 'FAILED'}")

design_ready = all(design_checks.values())
print("Responsive/accessibility baseline:", "READY" if design_ready else "NOT READY")
````
